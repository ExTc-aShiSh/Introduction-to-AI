```python
from collections import deque

def bfs(graph, start):
    visited = set() # Set to keep track of visited nodes
    queue = deque([start]) # Initialize the queue with the starting node

    visited.add(start)

    while queue:
        node = queue.popleft() # Dequeue a node
        print(node, end=" ") # Process the node (here, we just print it)

        # Explore all the neighbors of the current node
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor) # Mark the neighbor as visited
                queue.append(neighbor) # Enqueue the neighbor

# Example usage:
graph = {
    'electronic components': ['Active', 'Passive'],
    'Active': ['electronic components', 'Transistor', 'Diode'],
    'Passive': ['electronic components', 'Resister'],
    'Transistor': ['Active'],
    'Diode': ['Active'],
    'Resister': ['Passive']
}

start_node = 'electronic components'
print("BFS Traversal starting from node", start_node, ":")
bfs(graph, start_node)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
```

Now, let's create a `networkx` graph object from your dictionary and visualize it.

```python
G = nx.Graph(graph)

plt.figure(figsize=(15, 8))
pos = nx.spring_layout(G, k=50, iterations=200) # positions for all nodes
nx.draw_networkx_nodes(G, pos, node_size=2000, node_color='lightblue')
nx.draw_networkx_edges(G, pos, width=5, alpha=0.7, edge_color='orange')
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')
plt.title('Graph Visualization')
plt.axis('off') # Hide the axes
plt.show()
```
