```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()  # Initialize the visited set

    # Mark the current node as visited and print it
    visited.add(start)
    print(start, end=" ")


    # Recur for all the vertices adjacent to this node
    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


# Example usage:
graph = {
    'Sundernagar': ['Mandi', 'Chatri','Karsog','Shimla'],
    'Mandi': ['Sundernagar', 'Aut', 'Hamirpur','Sarkaghat','Jogindernagar'],
    'Sarkaghat' : ['Mandi','Dharampur','Hamirpur'],
    'Dharampur' : ['Sarkaghat','Sandhol','Jogindernagar'],
    'Jogindernagar': ['Mandi','Dharampur','Kangra'],
    'Hamirpur': ['Mandi', 'Shimla','Sarkaghat','Rangas','Sujanpur'],
    'Sujanpur': ['Hamirpur','Kangra','Sandhol'],
    'Sandhol': ['Dharampur','Sujanpur'],
    'Rangas' :['Hamirpur','Kangra'],
    'Kangra': ['Rangas','Jawali','Sujanpur','Jogindernagar'],
    'Jawali': ['Kangra'],
    'Aut': ['Mandi', 'Jalori','Kullu'],
    'Kullu' : ['Aut','Manali'],
    'Manali' : ['Kullu'],
    'Chatri': ['Sundernagar', 'Chaunti'],
    'Karsog': ['Sundernagar','Luhri','Churag'],
    'Churag': ['Karsog','Shimla'],
    'Shimla': ['Sundernagar', 'Luhri','Churag','Hamirpur'],
    'Jalori': ['Aut', 'Anni','Raghupur'],
    'Nigan': ['Chaunti', 'Anni','Raghupur'],
    'Raghupur': ['Nigan','Jalori'],
    'Chaunti': ['Nigan'],
    'Luhri' : ['Shimla','Karsog','Chaunti'],
    'Anni' : ['Jalori','Nigan',]

}


start_node = 'Sundernagar'
print("DFS Traversal starting from node", start_node, ":")
dfs(graph, start_node)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
# Convert the dictionary graph to a networkx graph
G = nx.Graph(graph)

# Visualize the graph
plt.figure(figsize=(20, 10))
pos = nx.spring_layout(G, k=0.5, iterations=50)  # Use spring layout for better visualization
nx.draw_networkx_nodes(G, pos, node_size=2000, node_color='skyblue')
nx.draw_networkx_edges(G, pos, edge_color='gray', alpha=0.7)
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')

plt.title('Graph Visualization', size=15)
plt.axis('off')  # Turn off the axis
plt.show()
```
